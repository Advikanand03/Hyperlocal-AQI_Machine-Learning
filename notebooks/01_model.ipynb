```python
import pandas as pd
import numpy as np
```

```python
file_path = "/Users/barkhaanand/Hyperlocal-AQI_Machine-Learning/data/raw/bangalore_caaqms/bangalore_caaqms_2025_combined.csv"
df = pd.read_csv(file_path)

df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
df = df[[
    "Timestamp",
    "station_name",
    "PM2.5 (µg/m³)",
    "AT (°C)",
    "RH (%)",
    "WS (m/s)",
    "WD (deg)"
]]
```

```python
df.columns = [
    "timestamp",
    "station",
    "pm25",
    "temperature",
    "humidity",
    "wind_speed",
    "wind_direction"
]
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"], errors="coerce")
```

```python
df["hour"] = df["timestamp"].dt.hour
df["month"] = df["timestamp"].dt.month
```

```python
clean_path = "../data/clean/bangalore_clean.csv"
df.to_csv(clean_path, index=False)

print("Clean dataset saved successfully!")
```
